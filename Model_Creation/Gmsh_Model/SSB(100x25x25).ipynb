## A simple rectangular slab

Dimentions: 100 x 25 x 25  (in mm)

```julia
import Gmsh: gmsh

# input parameters
const L = 100;           # Length
const B = 25;            # Breadth
const H = 25;             # Height
const h = 5;             # Mesh size

# Initializing Gmsh (Initialize the Gmsh API)
gmsh.initialize()

# Create a Model (For example, let's create a rectangle of length L and height H)
gmsh.model.add("Canvas")

# Define points
gmsh.model.geo.addPoint(0, 0, 0, h, 1)
gmsh.model.geo.addPoint(L, 0, 0, h, 2)
gmsh.model.geo.addPoint(L, B, 0, h, 3)
gmsh.model.geo.addPoint(0, B, 0, h, 4)

# Add lines
gmsh.model.geo.addLine(1, 2, 1)
gmsh.model.geo.addLine(2, 3, 2)
gmsh.model.geo.addLine(3, 4, 3)
gmsh.model.geo.addLine(4, 1, 4)

# Create a rectangular curve loop and surface
gmsh.model.geo.addCurveLoop([1, 2, 3, 4], 1)

# Create plane surface
s = gmsh.model.geo.addPlaneSurface([1])

# Extrude to create 3D 
# ext = gmsh.model.geo.extrude([(dim, tag)], dx, dy, dz)
ext = gmsh.model.geo.extrude([(2, s)], 0, 0, H)
print(ext)

# Entire 3D domain
gmsh.model.addPhysicalGroup(3, [1], 1)
gmsh.model.setPhysicalName(3, 1, "Domain")

# Face
gmsh.model.addPhysicalGroup(2, [s], 2)
gmsh.model.setPhysicalName(2, 2, "A")

# Face
gmsh.model.addPhysicalGroup(2, [26], 3)
gmsh.model.setPhysicalName(2, 3, "B")

# Face
gmsh.model.addPhysicalGroup(2, [13], 4)
gmsh.model.setPhysicalName(2, 4, "C")

# Face
gmsh.model.addPhysicalGroup(2, [17], 5)
gmsh.model.setPhysicalName(2, 5, "Hinge")

# Face
gmsh.model.addPhysicalGroup(2, [21], 6)
gmsh.model.setPhysicalName(2, 6, "D")

# Face
gmsh.model.addPhysicalGroup(2, [25], 7)
gmsh.model.setPhysicalName(2, 7, "Roller")


# Synchronize geometry 
gmsh.model.geo.synchronize()

# Mesh 
gmsh.model.mesh.generate(3)

# Create the folder (if it doesn't exist)
write_dir = joinpath(@__DIR__, "Model")
isdir(write_dir) || mkpath(write_dir)  

# Write mesh to file
gmsh.write(joinpath(write_dir, "SSB(100x25x25).msh"))

#Launch GUI 
gmsh.fltk.run()  

# Finalize Gmsh API
gmsh.finalize() 
```
