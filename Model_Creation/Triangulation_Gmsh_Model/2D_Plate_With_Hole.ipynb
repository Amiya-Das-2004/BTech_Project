## 2D Plate with a Hole

```julia
import Gmsh: gmsh

# input parameters
const L = 1000;           # Length
const B = 400;            # Breadth
const h = 25;             # Mesh size

x = (L/2);
y = (B/2);
r = 75;

a = x+r;
b = y+r;
c = x-r;
d = y-r;

println("Values = ", (a, b, c, d))

# Initializing Gmsh (Initialize the Gmsh API)
gmsh.initialize()

# Create a Model (For example, let's create a rectangle of length L and height H)
gmsh.model.add("Canvas")

# Define points
gmsh.model.geo.addPoint(0, 0, 0, h, 1)
gmsh.model.geo.addPoint(L, 0, 0, h, 2)
gmsh.model.geo.addPoint(L, B, 0, h, 3)
gmsh.model.geo.addPoint(0, B, 0, h, 4)

# Add lines
gmsh.model.geo.addLine(1, 2, 1)
gmsh.model.geo.addLine(2, 3, 2)
gmsh.model.geo.addLine(3, 4, 3)
gmsh.model.geo.addLine(4, 1, 4)

# Create a rectangular curve loop and surface
Rectangle = gmsh.model.geo.addCurveLoop([1, 2, 3, 4])

# Create circle arcs
gmsh.model.geo.addPoint(x, y,  0,  h,    5)     # Center
gmsh.model.geo.addPoint(a, y,  0,  h,    6)     # A1
gmsh.model.geo.addPoint(x, b,  0,  h,    7)     # A2
gmsh.model.geo.addPoint(c, y,  0,  h,    8)     # A3
gmsh.model.geo.addPoint(x, d,  0,  h,    9)     # A4

gmsh.model.geo.addCircleArc(6, 5, 7, 5)
gmsh.model.geo.addCircleArc(7, 5, 8, 6)
gmsh.model.geo.addCircleArc(8, 5, 9, 7)
gmsh.model.geo.addCircleArc(9, 5, 6, 8)

# Create circular curve loops
Circle = gmsh.model.geo.addCurveLoop([5, 6, 7, 8])

# Create plane surface
s = gmsh.model.geo.addPlaneSurface([1, 2])

# Domain
gmsh.model.addPhysicalGroup(2, [s], 1)
gmsh.model.setPhysicalName(2, 1, "Domain")

# line
gmsh.model.addPhysicalGroup(1, [5, 6, 7, 8], 2)
gmsh.model.setPhysicalName(1, 2, "Circle")

# Line
gmsh.model.addPhysicalGroup(1, [1], 3)
gmsh.model.setPhysicalName(1, 3, "A")

# Line
gmsh.model.addPhysicalGroup(1, [2], 4)
gmsh.model.setPhysicalName(1, 4, "B")

# Line
gmsh.model.addPhysicalGroup(1, [3], 5)
gmsh.model.setPhysicalName(1, 5, "C")

# Line
gmsh.model.addPhysicalGroup(1, [4], 6)
gmsh.model.setPhysicalName(1, 6, "D")

# Synchronize geometry 
gmsh.model.geo.synchronize()

# Mesh 
gmsh.model.mesh.generate(2)

# Create the folder (if it doesn't exist)
write_dir = joinpath(@__DIR__, "Model")
isdir(write_dir) || mkpath(write_dir)  

# Write mesh to file
gmsh.write(joinpath(write_dir, "2D_Plate_With_Hole.msh"))

#Launch GUI 
gmsh.fltk.run()  

# Finalize Gmsh API
gmsh.finalize() 
```
