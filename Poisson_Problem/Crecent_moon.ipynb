```julia
using Gridap
using GridapGmsh
using GridapEmbedded
using STLCutters
```

```julia
R  = 0.5
L  = 0.5*R
p1 = Point(0.0,0.0)
p2 = p1 + VectorValue(-L,L)

geo1 = disk(R,x0=p1)
geo2 = disk(R,x0=p2)
geo3 = setdiff(geo1,geo2)
```

```julia
t = 1.01
pmin = p1-t*R
pmax = p1+t*R

n = 30
partition = (n,n)
bgmodel = CartesianDiscreteModel(pmin,pmax,partition)
dp = pmax - pmin
```

```julia
cutgeo = cut(bgmodel,geo3)
```

```julia
Ω_act = Triangulation(cutgeo,ACTIVE)
```

```julia
Ω_bg = Triangulation(bgmodel)
writevtk(Ω_bg,"bg_trian")
writevtk(Ω_act,"act_trian")
```

```julia
Ω = Triangulation(cutgeo,PHYSICAL)
writevtk(Ω,"phys_trian")
```

```julia
order = 1
reffe = ReferenceFE(lagrangian,Float64,order)
Vstd = TestFESpace(Ω_act,reffe,conformity=:H1)
```

```julia
R = 1.0
center = Point(0.0,0.0,0.0)

geo = sphere(R, x0 = center)   # implicit level-set sphere

# Step 1: Background model
domain = (-2,2,-2,2,-2,2)
partition = (40,40,40)
bgmodel = CartesianDiscreteModel(domain, partition)

# Step 2: Cut with level-set
cutgeo = cut(bgmodel,geo)
```

```julia
Ω_act = Triangulation(cutgeo,ACTIVE)

Ω_bg = Triangulation(bgmodel)
```

```julia
writevtk(Ω_bg,"bg_trian")
writevtk(Ω_act,"act_trian")
```

```julia
Ω = Triangulation(cutgeo,PHYSICAL)
writevtk(Ω,"phys_trian")
```
