## Elliptical Interface Poisson Problem
This poblem was solved using FEM

##### Problem Statement

<div style="margin-left: 40px">

$ −∇·(𝛼ᵢ ∇𝑢) = 𝑓ᵢ   \qquad   \text{in }\,\, Ωᵢ \, ,\quad 𝑖 = 1, 2  $,

$ [𝑢] = 𝑔₀ = 0   \qquad  \text{on }\,\, Ω₀ $,

$ [𝑛 · 𝛼∇𝑢] = 𝑓₀ = 0  \qquad   \text{on }\,\, Ω₀ $, 

$ 𝑢 = 0   \qquad  \text{on }\,\, 𝜕Ω  $,

</div>

<div style="margin-left: 40px">

Weak Formulation

$ ∫_Ω (∇u)⋅(∇v)\,\,dx = ∫_Ω f⋅v\,\,dx  $

</div>

```julia
using Gridap
using GridapGmsh
```

```julia
# Call Model
model_name = "2D_Plate_With_Elliptical_Interface.msh"
model_file = joinpath(@__DIR__,"..", "Model_Creation", "Quadileteral_Gmsh_Model", "Model", model_name)
isfile(model_file) || error("Path does not exist: $model_file")
```

```julia
model = GmshDiscreteModel(model_file)
```

Hilbert Space defined using ReferenceFE and <br>
Drichlet boundary is defined where our test function (v) will be zero

```julia
order = 1
reffe = ReferenceFE(lagrangian, Float64, order)

# For Ω₁ outside domain
V0₁ = TestFESpace(model,reffe;
    conformity=:H1,
    dirichlet_tags=["∂Ω"],
    dirichlet_masks=[(true,true)])

# For Ω₂ outside domain
V0₂ = TestFESpace(model,reffe;
    conformity=:H1,
    dirichlet_tags=["Ω₀"],
    dirichlet_masks=[(true,true)])
```

```julia
# Approximation of the trial space where we choose the suitable solution function
g(x) = 0.0
U = TrialFESpace(V0₁,g)
```

```julia

degree = 2
Ω = Triangulation(model)
dΩ = Measure(Ω,degree)
```
